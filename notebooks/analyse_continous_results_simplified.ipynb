```python
def plot_metric_by_tiers_and_nodes(df, metric, fpath=None):
    """
    Plot a bar chart showing the average of the given metric vs number of tiers,
    with bars colored by number of nodes.
    
    Args:
        df: pandas DataFrame containing the data
        metric: str, name of metric column to plot
        fpath: str, optional filepath to save the plot
    """
    # Set publication theme
    set_pub_theme()

    if fpath:
        Path(fpath).parent.mkdir(parents=True, exist_ok=True)
        
    # Calculate mean metric by tiers and nodes
    mean_by_tiers_nodes = df.groupby(['num_tiers', 'nodes'])[metric].mean().reset_index()

    # Create figure and axis 
    fig, ax = plt.subplots(figsize=(12, 3), dpi=300)

    # Create grouped bar plot with cool to warm color palette
    bars = sns.barplot(data=mean_by_tiers_nodes, x='nodes', y=metric,
                      hue='num_tiers', errorbar=None, 
                      palette='coolwarm', width=0.9)
    
    # Add value labels on top of bars with larger font
    for i in bars.containers:
        bars.bar_label(i, fmt='%.2f', padding=3, fontsize=10)

    # Customize axis with explicit font sizes
    ax.set_xlabel('Number of Knowledge Tiers', fontsize=14)
    ax.set_ylabel(f'Average {metric.replace("_", " ").title() if metric not in ("SHD", "ECE", "Skeleton SHD") else metric}', fontsize=14)
    ax.set_title(f'{metric.replace("_", " ").title() if metric not in ("SHD", "ECE", "Skeleton SHD") else metric} vs Number of Knowledge Tiers by Number of Nodes', 
                 pad=20, fontsize=16)
    
    # Set tick label font sizes
    ax.tick_params(axis='both', labelsize=12)
    
    # Set legend font sizes
    plt.legend(title='Number\nof Nodes', bbox_to_anchor=(1, 1), loc='upper left',
              title_fontsize=12, fontsize=12)

    # Remove spines and adjust layout
    sns.despine()
    plt.tight_layout()
    
    if fpath:
        plt.savefig(fpath, bbox_inches='tight', dpi=300)
    
    plt.show()

for metric in metrics:
    plot_metric_by_tiers_and_nodes(df, metric, f'./plots/by_tiers_and_nodes/{metric}.png')
```
